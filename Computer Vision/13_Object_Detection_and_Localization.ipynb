```python
import tensorflow as tf
resnet_model = tf.keras.applications.ResNet50(
    weights='imagenet',
    include_top=False,
    input_shape=(224,224,3)
)
resnet_model.trainable = False
model_output = resnet_model.output
flattening = tf.keras.layers.Flatten()(model_output)
# classification head   --> This  head is to detect the class of the object of interest(single object) in the image Multi class Problem with 3 classes.
x1 = tf.keras.layers.Dense(128,activation='relu')(flattening)
x1 = tf.keras.layers.Dense(64,activation='relu')(x1)
x1 = tf.keras.layers.Dense(32,activation='relu')(x1)
output1 = tf.keras.layers.Dense(3,activation='softmax',name='classification_head')(x1)

# Regression head  --> This head is to detect the bx, by, bw, bh, bounding box of the object of interest(single object) in the image.
x2 = tf.keras.layers.Dense(128,activation='relu')(flattening)
x2 = tf.keras.layers.Dense(64,activation='relu')(x2)
x2 = tf.keras.layers.Dense(32,activation='relu')(x2)
output2 = tf.keras.layers.Dense(4,name='regression_head')(x2)

#Model 
model_output = tf.keras.models.Model(inputs=resnet_model.input,outputs=[output1,output2])


```

```python
lossess = {"classification_head":"categorical_crossentropy","regression_head":"mse"}
metrics = {"classification_head":"accuracy","regression_head":"mae"}
model_output.compile(optimizer='adam',loss=lossess,metrics=metrics)

#model.fit(x,[y1,y2])

```
